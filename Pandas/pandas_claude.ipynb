```python
'''
P5 — Derived column (feature engineering)
Create BMI_category using pd.cut (bins: <18.5 underweight, 18.5–25 normal, 25–30 overweight, >30 obese).
Then value_counts() on it.
'''

# Display the original dataset
df = pd.read_csv(r"D:\New python\practise claude\practisee\pandas data sets\diabetes.csv")
print("Original Dataset:")
print(df)

# 1. Define the bin edges and corresponding category labels
# Using -inf and inf ensures any values outside your explicit numbers (like a BMI of 12 or 55) are safely caught.
bins = [-float("inf"), 18.5, 25, 30, float("inf")]
labels = ["underweight", "normal", "overweight", "obese"]
# 2. Create the derived column using pd.cut
df["BMI_category"] = pd.cut(df["BMI"], bins=bins, labels=labels)

# 3. Get the distribution count for each category
print("BMI Category Counts:")
print(df["BMI_category"].value_counts())
```
