```python
'''P9 (starred) — groupby + filter (the real analyst move)
Mean rating per drugName, BUT only for drugs with at least 50 reviews 
(otherwise averages are noise). 
Chain: groupby ® agg (count & mean) ® filter count>=50 ® sort. 
This "filter to adequately-sampled groups" pattern is RWE bread and butter.
n Done when: you have a clean table of well-reviewed drugs ranked by rating.'''
```

```python
df = pd.read_csv(r"D:\New python\practise claude\practisee\pandas data sets\UCI ML Drug Review\drugsComTrain_raw.csv")

print("Orginal data Set")
print(df)

combine =df.groupby("drugName")["rating"].agg(["count","mean"])
print(combine)

review_50 = combine["count"] >= 50
print(review_50)

# 1. Filter the dataframe using your mask
filtered_drugs = combine[combine["count"] >= 50]

# 2. Sort by the 'mean' column from highest to lowest
final_ranking = filtered_drugs.sort_values(by="mean", ascending=False)

print(final_ranking.head(10))
```

```python
'''P10 — Dates & trend
Parse the date column (pd.to_datetime), extract year (.dt.year), and count reviews per year. Is the data
skewed to certain years?
n Done when: you have a reviews-per-year table.'''
```

```python
df = pd.read_csv(r"D:\New python\practise claude\practisee\pandas data sets\UCI ML Drug Review\drugsComTrain_raw.csv")

print("Orginal data Set")
print(df)

# 1. Convert the actual column in the dataframe
df["date"] = pd.to_datetime(df["date"],format = "%d-%b-%y")
print(df["date"])
# 2. Extract the year and save it to a new column
df["year"] = df["date"].dt.year

print(df["year"].value_counts())
```

```python
'''P11 (starred) — Aggregate & export (portfolio-worthy)
Build a summary table: for the top 20 conditions, show number of drugs reviewed,
total reviews, and mean rating.
Export to condition_summary.csv. This one goes in your GitHub README as a mini-analysis.
n Done when: the CSV exists and you've written 2–3 sentences interpreting it.'''
```

```python
df = pd.read_csv(r"D:\New python\practise claude\practisee\pandas data sets\UCI ML Drug Review\drugsComTrain_raw.csv")

print("Orginal data Set")
print(df)

# print("<-----Top 20 condition----->")
# top_20  =df["condition"].value_counts().head(20)

combine = df.groupby("condition").agg(
    total_reviews=("rating", "count"),
    mean_rating=("rating", "mean"),
    unique_drugs=("drugName", "nunique")
)
print(combine)

# Sort the entire summary table so the highest volume conditions are at the top
sorted_summary = combine.sort_values(by="total_reviews", ascending=False)

# Grab the top 20 most frequent conditions (this naturally drops the low-count HTML junk!)
top_20_conditions = sorted_summary.head(20)

print(top_20_conditions)

# Export the final clean summary to a CSV file as requested by the prompt
top_20_conditions.to_csv("condition_summary.csv")
```
